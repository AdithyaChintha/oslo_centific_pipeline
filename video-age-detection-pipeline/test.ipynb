```python
import cv2
from deepface import DeepFace
from datetime import timedelta

# === CONFIGURATION ===
video_path = "/Users/harshit.rajgarhia/Desktop/Sample Video/VID_20250720_152154_00_011.mp4"
frame_interval = 30  # Every 30 frames (~1s if 30fps)

cap = cv2.VideoCapture(video_path)
fps = cap.get(cv2.CAP_PROP_FPS)
frame_num = 0

print(f"Processing video: {video_path} with DeepFace")

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    if frame_num % frame_interval == 0:
        timestamp = str(timedelta(seconds=int(frame_num / fps)))
        try:
            result = DeepFace.analyze(frame, actions=["age", "gender"], enforce_detection=False)
            age = result[0]["age"]
            gender = result[0]["gender"]
            print(f"[{timestamp}] Age: {age}, Gender: {gender}")
        except Exception as e:
            print(f"[{timestamp}] DeepFace error: {e}")

    frame_num += 1

cap.release()
print("\n✅ Done with DeepFace.")
```

```python
import cv2
import numpy as np
from openvino.runtime import Core

# Paths to IR models (adjust if your folder layout differs)
FD_XML = "intel/face-detection-retail-0005/FP16/face-detection-retail-0005.xml"
AG_XML = "intel/age-gender-recognition-retail-0013/FP16/age-gender-recognition-retail-0013.xml"

ie = Core()

# Load models
fd_model = ie.read_model(FD_XML)
fd_compiled = ie.compile_model(fd_model, device_name="CPU")
fd_input = fd_compiled.inputs[0]
fd_output = fd_compiled.outputs[0]   # detection output: [1, 1, N, 7]

ag_model = ie.read_model(AG_XML)
ag_compiled = ie.compile_model(ag_model, device_name="CPU")
ag_input = ag_compiled.inputs[0]
ag_outputs = ag_compiled.outputs  # age + gender

# Helper: detect faces (Retail-0005 outputs [image_id, label, conf, x_min, y_min, x_max, y_max])
def detect_faces(bgr_img, conf_thr=0.6):
    h, w = bgr_img.shape[:2]
    n, c, ih, iw = fd_input.shape  # [1,3,300,300]
    resized = cv2.resize(bgr_img, (iw, ih))
    input_blob = resized.transpose(2, 0, 1)[None]  # NCHW
    res = fd_compiled([input_blob])[fd_output]     # shape [1,1,N,7]
    detections = []
    for det in res[0, 0]:
        conf = float(det[2])
        if conf < conf_thr:
            continue
        xmin = int(det[3] * w)
        ymin = int(det[4] * h)
        xmax = int(det[5] * w)
        ymax = int(det[6] * h)
        # clamp
        xmin, ymin = max(0, xmin), max(0, ymin)
        xmax, ymax = min(w - 1, xmax), min(h - 1, ymax)
        if xmax > xmin and ymax > ymin:
            detections.append((conf, (xmin, ymin, xmax, ymax)))
    return detections

# Helper: run age-gender on a cropped face
def predict_age_gender(bgr_face):
    # Get expected input shape dynamically (age-gender-0013 is typically [1,3,62,62])
    _, c, ih, iw = ag_input.shape
    face_resized = cv2.resize(bgr_face, (iw, ih))
    face_blob = face_resized.transpose(2, 0, 1)[None]  # NCHW
    outputs = ag_compiled([face_blob])

    # Model has two outputs:
    #  - "age_conv3": scalar age in [0,1], multiply by 100
    #  - "prob": gender probabilities [female, male]
    # Access by name if available; else fallback to list order
    if "age_conv3" in ag_compiled.output("age_conv3").get_any_name():
        age_output = outputs[ag_compiled.output("age_conv3")]
        gender_output = outputs[ag_compiled.output("prob")]
    else:
        # Fallback: assume outputs[0]=age, outputs[1]=gender
        age_output = outputs[ag_outputs[0]]
        gender_output = outputs[ag_outputs[1]]

    age = float(age_output.reshape(-1)[0]) * 100.0
    gender_probs = gender_output.reshape(-1)
    gender = "Male" if (len(gender_probs) > 1 and gender_probs[1] > gender_probs[0]) else "Female"
    return int(round(age)), gender, gender_probs

# --------- Run on an image ----------
image_path = "/Users/harshit.rajgarhia/Desktop/sample1.png"  # <- put your image here
img = cv2.imread(image_path)
assert img is not None, f"Cannot read image: {image_path}"

detections = detect_faces(img, conf_thr=0.6)

for i, (conf, (x1, y1, x2, y2)) in enumerate(detections, 1):
    # add a small margin around the face crop
    h, w = img.shape[:2]
    pad = int(0.1 * max(x2 - x1, y2 - y1))
    xa = max(0, x1 - pad)
    ya = max(0, y1 - pad)
    xb = min(w, x2 + pad)
    yb = min(h, y2 + pad)

    face_crop = img[ya:yb, xa:xb]
    age, gender, _ = predict_age_gender(face_crop)

    # draw results
    cv2.rectangle(img, (x1, y1), (x2, y2), (0, 255, 0), 2)
    label = f"{gender}, {age}"
    cv2.putText(img, label, (x1, max(0, y1 - 8)), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 0), 2)

# save result
out_path = "openvino_age_result.jpg"
cv2.imwrite(out_path, img)
print(f"Saved: {out_path}")
```

```python
import cv2
import numpy as np
from openvino.runtime import Core

FD_XML = "intel/face-detection-retail-0005/FP16/face-detection-retail-0005.xml"
AG_XML = "intel/age-gender-recognition-retail-0013/FP16/age-gender-recognition-retail-0013.xml"

ie = Core()

# Load models
fd_model = ie.read_model(FD_XML)
fd_compiled = ie.compile_model(fd_model, device_name="CPU")
fd_input = fd_compiled.inputs[0]
fd_output = fd_compiled.outputs[0]  # [1,1,N,7]

ag_model = ie.read_model(AG_XML)
ag_compiled = ie.compile_model(ag_model, device_name="CPU")
ag_input = ag_compiled.inputs[0]
ag_out0, ag_out1 = ag_compiled.outputs  # two outputs (order can vary by version)

def detect_faces(bgr_img, conf_thr=0.3):
    h, w = bgr_img.shape[:2]
    # Expect [1,3,300,300] for this model
    n, c, ih, iw = [int(x) for x in fd_input.shape]
    resized = cv2.resize(bgr_img, (iw, ih))
    # Some runtimes prefer float32
    input_blob = resized.transpose(2, 0, 1)[None].astype(np.float32)
    # Inference
    res_map = fd_compiled([input_blob])
    dets = res_map[fd_output]  # shape [1,1,N,7]
    detections = []
    max_conf = 0.0
    for det in dets[0, 0]:
        conf = float(det[2])
        max_conf = max(max_conf, conf)
        if conf < conf_thr:
            continue
        xmin = int(det[3] * w)
        ymin = int(det[4] * h)
        xmax = int(det[5] * w)
        ymax = int(det[6] * h)
        xmin, ymin = max(0, xmin), max(0, ymin)
        xmax, ymax = min(w - 1, xmax), min(h - 1, ymax)
        if xmax > xmin and ymax > ymin:
            detections.append((conf, (xmin, ymin, xmax, ymax)))
    print(f"[FaceDet] found={len(detections)}  max_conf={max_conf:.3f}")
    return detections

def predict_age_gender(bgr_face):
    # Typical input is [1,3,62,62]; query it to be safe
    _, c, ih, iw = [int(x) for x in ag_input.shape]
    face_resized = cv2.resize(bgr_face, (iw, ih))
    face_blob = face_resized.transpose(2, 0, 1)[None].astype(np.float32)
    outputs = ag_compiled([face_blob])

    # The two outputs are age (scalar [0,1]) and gender probs [female, male]
    # Figure out which is which by shape
    out_a = outputs[ag_out0]
    out_b = outputs[ag_out1]
    # one is scalar-like, the other is length-2 vector
    if np.prod(out_a.shape) == 1:
        age_output = out_a
        gender_output = out_b
    else:
        age_output = out_b
        gender_output = out_a

    age = float(age_output.reshape(-1)[0]) * 100.0
    gender_probs = gender_output.reshape(-1)
    gender = "Male" if (len(gender_probs) > 1 and gender_probs[1] > gender_probs[0]) else "Female"
    return int(round(age)), gender, gender_probs

# --------- Run on an image ----------
image_path = "/Users/harshit.rajgarhia/Desktop/sample2.png"
img = cv2.imread(image_path)
assert img is not None, f"Cannot read image: {image_path}"

detections = detect_faces(img, conf_thr=0.3)

for i, (conf, (x1, y1, x2, y2)) in enumerate(detections, 1):
    h, w = img.shape[:2]
    pad = int(0.12 * max(x2 - x1, y2 - y1))
    xa = max(0, x1 - pad); ya = max(0, y1 - pad)
    xb = min(w, x2 + pad); yb = min(h, y2 + pad)

    face_crop = img[ya:yb, xa:xb]
    if face_crop.size == 0:
        continue

    age, gender, _ = predict_age_gender(face_crop)

    # Draw detection with confidence first (to verify boxes are actually drawn)
    cv2.rectangle(img, (x1, y1), (x2, y2), (0, 255, 0), 2)
    cv2.putText(img, f"{conf:.2f}", (x1, y1 - 24), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 0), 2)
    # Then age/gender label
    cv2.putText(img, f"{gender}, {age}", (x1, y1 - 4), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 0), 2)

out_path = "openvino_age_result.jpg"
cv2.imwrite(out_path, img)
print(f"Saved: {out_path}")
```

```python
import cv2
import numpy as np
from openvino.runtime import Core

# 1) Use a more forgiving detector
FD_XML = "./models/intel/face-detection-adas-0001/FP16/face-detection-adas-0001.xml"
AG_XML = "./models/intel/age-gender-recognition-retail-0013/FP16/age-gender-recognition-retail-0013.xml"

ie = Core()

# Load models
fd_model = ie.read_model(FD_XML)
fd_compiled = ie.compile_model(fd_model, device_name="CPU")
fd_input = fd_compiled.inputs[0]
fd_output = fd_compiled.outputs[0]  # [1,1,N,7] format

ag_model = ie.read_model(AG_XML)
ag_compiled = ie.compile_model(ag_model, device_name="CPU")
ag_input = ag_compiled.inputs[0]
ag_out0, ag_out1 = ag_compiled.outputs  # two outputs

def detect_faces(bgr_img, conf_thr=0.25):
    h, w = bgr_img.shape[:2]
    n, c, ih, iw = [int(x) for x in fd_input.shape]  # e.g. [1,3,384,672] for ADAS, or 300x300
    resized = cv2.resize(bgr_img, (iw, ih))
    inp = resized.transpose(2, 0, 1)[None].astype(np.float32)
    res_map = fd_compiled([inp])
    dets = res_map[fd_output]  # [1,1,N,7]

    out = []
    max_conf = 0.0
    for det in dets[0, 0]:
        conf = float(det[2])
        max_conf = max(max_conf, conf)
        if conf < conf_thr:
            continue
        xmin = int(det[3] * w)
        ymin = int(det[4] * h)
        xmax = int(det[5] * w)
        ymax = int(det[6] * h)
        xmin, ymin = max(0, xmin), max(0, ymin)
        xmax, ymax = min(w - 1, xmax), min(h - 1, ymax)
        if xmax > xmin and ymax > ymin:
            out.append((conf, (xmin, ymin, xmax, ymax)))
    print(f"[FaceDet] found={len(out)} max_conf={max_conf:.3f}")
    return out

def square_expand(x1, y1, x2, y2, img_w, img_h, scale=1.5):
    """Make the face box square and enlarge it by `scale`,
    centered on the original box, clamped to image bounds.
    """
    w = x2 - x1
    h = y2 - y1
    cx = x1 + w / 2.0
    cy = y1 + h / 2.0
    side = max(w, h) * scale

    nx1 = int(round(cx - side / 2))
    ny1 = int(round(cy - side / 2))
    nx2 = int(round(cx + side / 2))
    ny2 = int(round(cy + side / 2))

    nx1, ny1 = max(0, nx1), max(0, ny1)
    nx2, ny2 = min(img_w - 1, nx2), min(img_h - 1, ny2)
    return nx1, ny1, nx2, ny2

def predict_age_gender(face_bgr):
    # Model expects [1,3,62,62]; query to be safe
    _, _, ih, iw = [int(x) for x in ag_input.shape]
    face_resized = cv2.resize(face_bgr, (iw, ih))
    blob = face_resized.transpose(2, 0, 1)[None].astype(np.float32)
    outputs = ag_compiled([blob])

    a = outputs[ag_out0]
    b = outputs[ag_out1]
    # Identify scalar vs probs
    if np.prod(a.shape) == 1:
        age_out, prob_out = a, b
    else:
        age_out, prob_out = b, a

    age = float(age_out.reshape(-1)[0]) * 100.0
    probs = prob_out.reshape(-1)
    gender = "Male" if (len(probs) > 1 and probs[1] > probs[0]) else "Female"
    return int(round(age)), gender, probs

# -------- Run ----------
image_path = "/Users/harshit.rajgarhia/Desktop/sample1.png"
img = cv2.imread(image_path)
assert img is not None, f"Cannot read image: {image_path}"

h, w = img.shape[:2]
dets = detect_faces(img, conf_thr=0.25)

for conf, (x1, y1, x2, y2) in dets:
    # 2) Better crop: enlarge + square so the 62x62 model sees full head/forehead
    sx1, sy1, sx2, sy2 = square_expand(x1, y1, x2, y2, w, h, scale=1.6)
    crop = img[sy1:sy2, sx1:sx2]
    if crop.size == 0:
        continue

    age, gender, _ = predict_age_gender(crop)

    # Draw original box for reference
    cv2.rectangle(img, (x1, y1), (x2, y2), (0, 255, 0), 2)
    # Label
    cv2.putText(img, f"{gender}, {age}  ({conf:.2f})",
                (x1, max(0, y1 - 8)), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 0), 2)

out_path = "openvino_age_result_better.jpg"
cv2.imwrite(out_path, img)
print(f"Saved: {out_path}")
```
