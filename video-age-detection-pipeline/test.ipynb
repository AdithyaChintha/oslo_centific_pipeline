```python
import cv2
from deepface import DeepFace
from datetime import timedelta

# === CONFIGURATION ===
video_path = "/Users/harshit.rajgarhia/Desktop/Sample Video/VID_20250720_152154_00_011.mp4"
frame_interval = 30  # Every 30 frames (~1s if 30fps)

cap = cv2.VideoCapture(video_path)
fps = cap.get(cv2.CAP_PROP_FPS)
frame_num = 0

print(f"Processing video: {video_path} with DeepFace")

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    if frame_num % frame_interval == 0:
        timestamp = str(timedelta(seconds=int(frame_num / fps)))
        try:
            result = DeepFace.analyze(frame, actions=["age", "gender"], enforce_detection=False)
            age = result[0]["age"]
            gender = result[0]["gender"]
            print(f"[{timestamp}] Age: {age}, Gender: {gender}")
        except Exception as e:
            print(f"[{timestamp}] DeepFace error: {e}")

    frame_num += 1

cap.release()
print("\n✅ Done with DeepFace.")
```
